## 1. 启动 FastAPI 服务

启动后会自动打开浏览器。**服务会一直跑，看完点 ■ 中断内核。**

```
import subprocess, webbrowser, time

proc = subprocess.Popen(["python", "app.py"])
time.sleep(20)
webbrowser.open("http://localhost:8000")
print("已尝试打开 http://localhost:8000")
print("如果页面没出来，等几秒手动打开（启动要读 5.4MB Excel，约 20 秒）")
```

## 2. 直接请求各个页面（验证服务正常）

```
import urllib.request

for p in ["/", "/module/food", "/module/game", "/module/retention", "/module/loan", "/agent"]:
    try:
        with urllib.request.urlopen("http://localhost:8000" + p, timeout=120) as r:
            print(f"{p:22} HTTP {r.status}  {len(r.read()):>8} bytes")
    except Exception as e:
        print(f"{p:22} 失败：{e}")
```

## 3. 单独看贷款风控模块的核心数字

```
import urllib.request, re
html = urllib.request.urlopen("http://localhost:8000/module/loan", timeout=120).read().decode()
for m in re.findall(r"(AUC|随机森林|逻辑回归)[^<]{0,30}", html)[:10]:
    print(m)
```

## 停止服务

```
proc.terminate()
print("服务已停止")
```
